# Poster Visualizations
### Utah Legislature Bill Outcome Analysis — DS 2500 Final Project

This notebook generates the five visualizations for the poster:
1. Subject × Party pass rate heatmap
2. Pass rate by primary subject bar chart
3. Decision tree (simplified)
4. Random forest feature importance
5. Economic indicator effects (logistic regression interaction coefficients)

All figures are saved as high-resolution PNGs for import into Canva.

## Setup

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Poster-friendly styling: larger fonts, cleaner look, high DPI
sns.set_style('whitegrid')
plt.rcParams.update({
    'font.size': 14,
    'axes.titlesize': 18,
    'axes.labelsize': 15,
    'xtick.labelsize': 13,
    'ytick.labelsize': 13,
    'legend.fontsize': 13,
    'figure.dpi': 200,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'axes.spines.top': False,
    'axes.spines.right': False,
})
```

```
# Load data and prepare analysis sample
df_primary = pd.read_csv('df_primary.csv')
df_bills = pd.read_csv('df_bills.csv')[['bill_id', 'session_year', 'party', 'party1']]
df_raw = df_primary.merge(df_bills, on=['bill_id', 'session_year'], how='left')

df = df_raw[df_raw['floorspon1'] == 1].copy()
df = df.dropna(subset=['party1']).copy()
df['party1'] = df['party1'].astype(int)
df['house_origin'] = (df['originating_chamber'] == 'House').astype(int)
df['subjects_changed_bin'] = df['subjects_changed'].astype(int)

SUBJECT_COLS = [
    'agri_food', 'arts_cult_rec', 'bus_comm', 'civ_rights_libs', 'crim_justice',
    'econ_dev', 'ed_k12', 'elect_gov', 'energy', 'env_nr', 'fam_ss',
    'health_care', 'higher_ed', 'house_land_use', 'labor_ee', 'local_gov',
    'no_subject', 'pub_safe_emerg', 'retire_ben', 'tax_fin', 'tech_data', 'transpo'
]
BILL_COVARIATES = ['committee1', 'money1', 'num_sub_new', 'money_sum_real', 'party1']
ECON_VARS = ['real_gdp', 'rental_vacancy_rate', 'poverty_rate', 'inflation']

# Human-readable subject names for poster
SUBJECT_LABELS = {
    'agri_food': 'Agriculture & Food',
    'arts_cult_rec': 'Arts, Culture & Rec',
    'bus_comm': 'Business & Commerce',
    'civ_rights_libs': 'Civil Rights',
    'crim_justice': 'Criminal Justice',
    'econ_dev': 'Economic Development',
    'ed_k12': 'K-12 Education',
    'elect_gov': 'Elections & Gov Ops',
    'energy': 'Energy',
    'env_nr': 'Environment & Nat. Res.',
    'fam_ss': 'Family & Social Svcs',
    'health_care': 'Healthcare',
    'higher_ed': 'Higher Education',
    'house_land_use': 'Housing & Land Use',
    'labor_ee': 'Labor & Employment',
    'local_gov': 'Local Government',
    'no_subject': 'No Subject',
    'pub_safe_emerg': 'Public Safety',
    'retire_ben': 'Retirement & Benefits',
    'tax_fin': 'Taxes & Finance',
    'tech_data': 'Technology & Data',
    'transpo': 'Transportation'
}

COVARIATE_LABELS = {
    'committee1': 'Came from Committee',
    'money1': 'Money Appropriated',
    'num_sub_new': '# of Substitutes',
    'money_sum_real': 'Total $ Appropriated',
    'party1': 'Republican Sponsor',
    'house_origin': 'House Origin',
    'subjects_changed_bin': 'Subjects Changed',
}

print(f"Analysis sample: {len(df)} floor-sponsored bills")
print(f"Pass rate: {df['passed'].mean():.1%}")
```

## Visualization 1 — Subject × Party Heatmap

```
subject_party = []
for col in SUBJECT_COLS:
    for party in ['Rep', 'Dem']:
        subset = df[(df[col] == 1) & (df['party'] == party)]
        if len(subset) > 5:
            subject_party.append({'subject': SUBJECT_LABELS[col], 'party': party,
                                  'pass_rate': subset['passed'].mean(), 'n': len(subset)})

sp_df = pd.DataFrame(subject_party)
sp_pivot = sp_df.pivot(index='subject', columns='party', values='pass_rate')
sp_pivot['gap'] = sp_pivot['Rep'] - sp_pivot['Dem']
sp_pivot = sp_pivot.sort_values('gap')

fig, ax = plt.subplots(figsize=(8, 10))
sns.heatmap(sp_pivot[['Dem', 'Rep']], annot=True, fmt='.0%', cmap='RdYlGn',
            center=df['passed'].mean(), linewidths=1, linecolor='white',
            cbar_kws={'label': 'Pass Rate'},
            annot_kws={'size': 13, 'weight': 'bold'}, ax=ax)
ax.set_title('Pass Rate by Subject & Sponsor Party', pad=15, fontweight='bold')
ax.set_xlabel('Sponsor Party', fontweight='bold')
ax.set_ylabel('')
ax.tick_params(axis='y', rotation=0)

plt.tight_layout()
plt.savefig('poster_01_subject_party_heatmap.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print("Saved: poster_01_subject_party_heatmap.png")
```

## Visualization 2 — Pass Rate by Primary Subject

```
pass_rates = {}
for col in SUBJECT_COLS:
    subset = df[df[col] == 1]
    if len(subset) > 0:
        pass_rates[SUBJECT_LABELS[col]] = {
            'pass_rate': subset['passed'].mean(),
            'count': len(subset)
        }

pr_df = pd.DataFrame(pass_rates).T.sort_values('pass_rate', ascending=True)
overall = df['passed'].mean()

fig, ax = plt.subplots(figsize=(10, 9))
colors = ['#27ae60' if r > overall else '#c0392b' for r in pr_df['pass_rate']]
bars = ax.barh(pr_df.index, pr_df['pass_rate'], color=colors,
               edgecolor='white', linewidth=1.5)

ax.axvline(overall, color='black', linestyle='--', linewidth=2,
           label=f'Overall: {overall:.0%}', alpha=0.7)
ax.set_xlabel('Pass Rate', fontweight='bold')
ax.set_title('Bill Pass Rate by Policy Area', pad=15, fontweight='bold')
ax.legend(loc='lower right', frameon=True, edgecolor='black')
ax.set_xlim(0, 1.05)
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.0%}'))

# Annotate with bill counts
for bar, (_, row) in zip(bars, pr_df.iterrows()):
    ax.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
            f'n={int(row["count"])}', va='center', fontsize=11, color='#555')

plt.tight_layout()
plt.savefig('poster_02_pass_rate_by_subject.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print("Saved: poster_02_pass_rate_by_subject.png")
```

## Visualization 3 — Decision Tree (Simplified)

```
feature_cols = SUBJECT_COLS + BILL_COVARIATES + ['house_origin', 'subjects_changed_bin']
y_full = df['passed']

# Shorter readable feature names for tree labels
readable_names = []
for f in feature_cols:
    if f in SUBJECT_LABELS:
        readable_names.append(SUBJECT_LABELS[f])
    elif f in COVARIATE_LABELS:
        readable_names.append(COVARIATE_LABELS[f])
    else:
        readable_names.append(f)

# Depth 3 tree — cleaner for poster
dt = DecisionTreeClassifier(max_depth=3, min_samples_leaf=100,
                             random_state=42, class_weight='balanced')
dt.fit(df[feature_cols], y_full)

fig, ax = plt.subplots(figsize=(20, 10))
plot_tree(dt, feature_names=readable_names,
          class_names=['Failed', 'Passed'],
          filled=True, rounded=True, fontsize=13,
          proportion=True, impurity=False, precision=2, ax=ax)
ax.set_title('Decision Tree: Predicting Bill Passage', pad=15, fontsize=20, fontweight='bold')

plt.tight_layout()
plt.savefig('poster_03_decision_tree.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print("Saved: poster_03_decision_tree.png")
```

## Visualization 4 — Random Forest Feature Importance

```
rf = RandomForestClassifier(n_estimators=200, max_depth=10, min_samples_leaf=20,
                             random_state=42, n_jobs=-1, class_weight='balanced')
rf.fit(df[feature_cols], y_full)

imp_df = pd.DataFrame({
    'Feature': readable_names,
    'Importance': rf.feature_importances_,
    'raw': feature_cols
}).sort_values('Importance', ascending=True)

# Color by variable type: procedural (blue), party (orange), subject (gray)
procedural = ['committee1', 'money1', 'num_sub_new', 'money_sum_real',
              'house_origin', 'subjects_changed_bin']
party_vars = ['party1']

colors = []
for raw in imp_df['raw']:
    if raw in party_vars:
        colors.append('#e67e22')
    elif raw in procedural:
        colors.append('#2980b9')
    else:
        colors.append('#7f8c8d')

fig, ax = plt.subplots(figsize=(10, 10))
ax.barh(imp_df['Feature'], imp_df['Importance'], color=colors,
        edgecolor='white', linewidth=1.5)

ax.set_xlabel('Feature Importance', fontweight='bold')
ax.set_title('What Predicts Bill Passage?', pad=15, fontweight='bold')

# Legend
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#2980b9', label='Procedural'),
    Patch(facecolor='#e67e22', label='Party'),
    Patch(facecolor='#7f8c8d', label='Policy Subject'),
]
ax.legend(handles=legend_elements, loc='lower right', frameon=True, edgecolor='black')

plt.tight_layout()
plt.savefig('poster_04_rf_importance.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print("Saved: poster_04_rf_importance.png")
```

## Visualization 5 — Economic Indicator Effects

Two complementary views of how economic context influences bill passage:

**5a**: Model comparison bar chart showing that adding ACS economic indicators barely improves predictive power

**5b**: The strongest economic × subject interaction coefficients, showing that while macro-level economic effects are weak, economic context *does* modulate specific policy areas

```
# Prepare data for logistic regression models
X_full_cols = SUBJECT_COLS + BILL_COVARIATES + ['house_origin', 'subjects_changed_bin']
df_econ = df.dropna(subset=ECON_VARS).copy()
y_econ = df_econ['passed'].copy()

X1 = df_econ[X_full_cols].copy()
X2 = df_econ[X_full_cols + ECON_VARS].copy()

interaction_subjects = ['health_care', 'fam_ss', 'ed_k12', 'crim_justice', 'tax_fin']
X3 = X2.copy()
for econ in ECON_VARS:
    for subj in interaction_subjects:
        X3[f'{econ}_x_{subj}'] = X3[econ] * X3[subj]
for econ in ECON_VARS:
    X3[f'{econ}_x_party1'] = X3[econ] * X3['party1']

from sklearn.model_selection import cross_val_score, StratifiedKFold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scaler = StandardScaler()

aucs = {}
for name, X in [('Bill Features\nOnly', X1),
                 ('+ Economic\nIndicators', X2),
                 ('+ Interactions', X3)]:
    X_scaled = scaler.fit_transform(X)
    lr = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
    auc = cross_val_score(lr, X_scaled, y_econ, cv=cv, scoring='roc_auc')
    aucs[name] = (auc.mean(), auc.std())

# Plot 5a: Model comparison
fig, ax = plt.subplots(figsize=(9, 6))
names = list(aucs.keys())
means = [aucs[n][0] for n in names]
stds = [aucs[n][1] for n in names]

bar_colors = ['#7f8c8d', '#3498db', '#2c3e50']
bars = ax.bar(names, means, yerr=stds, color=bar_colors,
              edgecolor='white', linewidth=2, capsize=8,
              error_kw={'linewidth': 2, 'ecolor': '#333'})

ax.axhline(0.5, color='red', linestyle=':', linewidth=2,
           label='Random baseline (0.5)', alpha=0.7)
ax.set_ylabel('ROC-AUC', fontweight='bold')
ax.set_title('Do Economic Indicators Improve Predictions?', pad=15, fontweight='bold')
ax.set_ylim(0.45, 0.70)
ax.legend(loc='lower right', frameon=True, edgecolor='black')

# Annotate bars with values
for bar, val in zip(bars, means):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.008,
            f'{val:.3f}', ha='center', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('poster_05a_econ_model_comparison.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print("Saved: poster_05a_econ_model_comparison.png")
```

```
# Plot 5b: Economic interaction coefficients (Model 3)
X_scaled_3 = scaler.fit_transform(X3)
lr3 = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
lr3.fit(X_scaled_3, y_econ)

# Get interaction term coefficients only
interaction_mask = ['_x_' in col for col in X3.columns]
interaction_coefs = pd.DataFrame({
    'term': X3.columns[interaction_mask],
    'coef': lr3.coef_[0][interaction_mask]
})

# Readable labels
econ_labels = {
    'real_gdp': 'Real GDP',
    'inflation': 'Inflation',
    'poverty_rate': 'Poverty Rate',
    'rental_vacancy_rate': 'Rental Vacancy'
}
subj_labels_short = {
    'health_care': 'Healthcare',
    'fam_ss': 'Family & Social Svcs',
    'ed_k12': 'K-12 Education',
    'crim_justice': 'Criminal Justice',
    'tax_fin': 'Taxes & Finance',
    'party1': 'Republican'
}

def pretty(term):
    econ, other = term.split('_x_')
    e = econ_labels.get(econ, econ)
    o = subj_labels_short.get(other, other)
    return f'{e} × {o}'

interaction_coefs['label'] = interaction_coefs['term'].apply(pretty)
interaction_coefs = interaction_coefs.sort_values('coef')

fig, ax = plt.subplots(figsize=(11, 10))
colors = ['#27ae60' if c > 0 else '#c0392b' for c in interaction_coefs['coef']]
ax.barh(interaction_coefs['label'], interaction_coefs['coef'],
        color=colors, edgecolor='white', linewidth=1.5)
ax.axvline(0, color='black', linewidth=1.2)
ax.set_xlabel('Coefficient (standardized log-odds)', fontweight='bold')
ax.set_title('Economic Context × Policy Area Effects', pad=15, fontweight='bold')

plt.tight_layout()
plt.savefig('poster_05b_econ_interactions.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print("Saved: poster_05b_econ_interactions.png")
```
